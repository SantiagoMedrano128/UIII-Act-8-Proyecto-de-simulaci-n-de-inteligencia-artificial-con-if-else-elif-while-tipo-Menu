####  Utiliza una lista de diccionarios para gestionar la base de datos de los oficiales y emplea lógica de inteligencia artificial básica para la toma de decisiones tácticas.

```python
# --- SIMULADOR DE IA: SISTEMA DE COMANDO POLICIAL 2026 ---

# Se inicializa la lista de policía como una lista vacía [2, 5]
policia = []
# Se inicializa el diccionario policias como un diccionario vacío [3, 6]
lospolicias = {}

# Inicio del ciclo infinito para el menú interactivo [1, 4]
while True:
    print("\n" + "=".center(50, "="))
    print("   CENTRO DE INTELIGENCIA TÁCTICA POLICIAL   ")
    print("=".center(50, "="))
    print("1. Registrar nuevo oficial")
    print("2. Ver lista de oficiales (Despliegue)")
    print("3. Analizador de Incidentes IA (Lógica Táctica)")
    print("4. Salir del sistema")

    # Captura de la opción del usuario [7, 8]
    opcion = input("\nSeleccione una opción (1-4): ")

    # Estructura de decisión para la Opción 1 [9, 10]
    if opcion == "1":
        print("\n[MÓDULO DE REGISTRO INTERACTIVO]")
        nombre = input("Nombre del oficial: ")
        # Se ajusta el formato del texto ingresado [11]
        rango = input("Rango (Cadete / Sargento / Capitan): ").capitalize()
        edad = int(input("Edad del oficial: "))

        # Se llena el diccionario 'policias' con los datos capturados [3, 12]
        policias = {"nombre": nombre, "rango": rango, "edad": edad}
        # Se agrega el registro a la lista principal [13, 14]
        policia.append(policias)
        print(f"IA: Oficial {nombre} almacenado en la base de datos.")

    # Estructura de decisión para la Opción 2 [9, 10]
    elif opcion == "2":
        print("\n[LISTADO DE PERSONAL EN SERVICIO]")
        # Verificación de lista vacía mediante longitud [15, 16]
        if len(policia) == 0:
            print("IA: No hay oficiales registrados actualmente.")
        else:
            # Ciclo for para recorrer la lista de diccionarios [17, 18]
            for oficial in policia:
                # Acceso a los valores mediante las llaves del diccionario [12, 19]
                print(f"Oficial: {oficial['nombre']} | Rango: {oficial['rango']} | Edad: {oficial['edad']}")

    # Estructura de decisión para la Opción 3 (Lógica de IA) [9, 20]
    elif opcion == "3":
        print("\n[MÓDULO DE ANÁLISIS DE RIESGO IA]")
        distancia = float(input("Distancia del incidente (km): "))
        gravedad = int(input("Nivel de amenaza detectado (1-10): "))
        tipo = input("¿El crimen es un Robo o una Emergencia? ").lower()

        # IA Lógica: Uso de operadores relacionales y lógicos para decidir [21, 22]
        if gravedad >= 8 and (tipo == "robo" or tipo == "emergencia"):
            print("IA-STATUS: !!! ALERTA ROJA !!! Despliegue de unidades SWAT.")
            # Cálculo aritmético para simular el tiempo de respuesta [23, 24]
            tiempo = (distancia / 60) * 60
            print(f"IA-CÁLCULO: Tiempo de llegada estimado: {tiempo:.1f} minutos.")

        elif gravedad >= 4 or distancia <= 3:
            print("IA-STATUS: ALERTA AMARILLA. Enviar patrulla de zona.")

        else:
            print("IA-STATUS: Situación bajo control. Mantener vigilancia remota.")

    # Opción para terminar la ejecución del programa [1, 25]
    elif opcion == "4":
        print("\nIA: Desconectando sistema de seguridad... ¡Hasta pronto!")
        break

    # Validación de entradas no reconocidas [26, 27]
    else:
        print("\nIA: Error de comando. Introduzca una opción válida (1-4).")

print("--- SESIÓN FINALIZADA ---")
```
