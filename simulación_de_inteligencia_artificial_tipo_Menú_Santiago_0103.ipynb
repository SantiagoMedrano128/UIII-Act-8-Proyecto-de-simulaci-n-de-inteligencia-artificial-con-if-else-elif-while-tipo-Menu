## Juez : mazo
#### Este programa simula un Sistema de Justicia Virtual (IA-Juez), Utiliza una estructura interactiva basada en un menú, procesando decisiones judiciales y el uso simbólico del mazo mediante condicionales, operadores y un ciclo de control.
![link text](https://raw.githubusercontent.com/SantiagoMedrano128/UIII-Act-8-Proyecto-de-simulaci-n-de-inteligencia-artificial-con-if-else-elif-while-tipo-Menu/refs/heads/main/infografiasimulacionIAtipomenuSantiago0103.png)

```python
# --- SIMULADOR DE IA: JUEZ VIRTUAL Y SENTENCIAS ---

print("INICIALIZANDO IA DE CORTE DE JUSTICIA...")
print("---------------------------------------")

# Variable para mantener el programa funcionando
sistema_activo = "si"

# Ciclo principal para la interactividad
while sistema_activo == "si" or sistema_activo == "SI":
    print("\n--- MENÚ DEL TRIBUNAL IA ---")
    print("1. Dictar Sentencia (IA Juez)")
    print("2. Acción de Orden (Usar Mazo)")
    print("3. Estado del Tribunal")
    print("4. Salir")

    opcion = input("Seleccione una acción (1-4): ")

    # Lógica del Juez Virtual para dictar sentencias
    if opcion == "1":
        print("\n[MÓDULO JUEZ SELECCIONADO]")
        delito = input("Tipo de caso (Robo / Infracción): ").lower()
        gravedad = int(input("Nivel de gravedad del caso (1-10): "))
        edad = int(input("Edad del acusado: "))

        # Uso de operadores lógicos (and, or) y relacionales (>=, <)
        if edad < 18:
            print("IA-JUEZ: El acusado es menor de edad. Se remite a tribunal juvenil.")
        elif delito == "robo" and gravedad >= 7:
            print("IA-JUEZ: Sentencia Dictada: CULPABLE. Pena máxima sugerida.")
        elif delito == "robo" or delito == "infraccion":
            print("IA-JUEZ: Sentencia Dictada: MULTA ECONÓMICA. Gravedad moderada.")
        else:
            print("IA-JUEZ: Evidencia insuficiente para un veredicto automático.")

    # Lógica para la acción del Mazo
    elif opcion == "2":
        print("\n[MÓDULO MAZO DE JUSTICIA]")
        golpes = int(input("¿Cuántos golpes dará el Juez con el mazo? (1-5): "))

        if golpes > 0:
            # Operador aritmético para calcular el "nivel de orden"
            nivel_orden = golpes * 20
            print(f"IA-MAZO: ¡ORDEN EN LA SALA! Nivel de silencio alcanzado: {nivel_orden}%")

            if golpes >= 3:
                print("IA-MAZO: El sonido fue fuerte. La sala está en absoluto silencio.")
            else:
                print("IA-MAZO: El sonido fue leve. Algunos murmullos continúan.")
        else:
            print("IA-MAZO: El mazo no ha sido utilizado.")

    # Opción de Estado del Sistema
    elif opcion == "3":
        print("\n--- REPORTE TÉCNICO DE LA IA ---")
        # Operador de comparación simple
        print("Conexión con Base de Datos: Estable")
        print("Algoritmo de Ética: Activo")
        print("Carga de Procesamiento: Normal")

    # Opción para salir del programa y romper el ciclo
    elif opcion == "4":
        sistema_activo = "no"
        print("Cerrando sesión del Juez Virtual... Justicia procesada.")

    # Validación de entrada para el menú
    else:
        print("Opción inválida. El Juez no reconoce ese comando.")

# Mensaje final al terminar la ejecución
print("\n--- FIN DE LA SIMULACIÓN ---")
```
